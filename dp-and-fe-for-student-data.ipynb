```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("student.csv")
df_before = pd.read_csv("student.csv")
```

```python
df
```

```python
#ใช้วิธีแทนค่าว่างเปล่าด้วยการใช้ค่าเฉลี่ย
mean_age = df['อายุ'].mean()
df['อายุ'] = df['อายุ'].fillna(mean_age).astype(int)
df
```

```python
#ใช้วิธีแทนค่าว่างเปล่าด้วยค่าที่พบบ่อย
df['คณะ'].fillna(df['คณะ'].mode()[0], inplace=True)
df
```

```python
def standardize_faculty(name):
    name = name.lower()
    if name in ['วิทยาศาสตร์', 'วิทย์', 'science', 'คณะวิทยาศาสตร์','sci']:
        return 'วิทยาศาสตร์'
    elif name in ['ศิลปศาสตร์', 'ศิลป์', 'arts','liberal arts']:
        return 'ศิลปศาสตร์'
    elif name in ['วิศวฯ','วิศวกรรม','engineering','วิศวะ']:
        return 'วิศวกรรมศาสตร์'
    else:
        return name

df['คณะ'] = df['คณะ'].apply(standardize_faculty)
df['คณะ_มาตรฐาน'] = df['คณะ'].apply(standardize_faculty)
df
```

```python
#ใช้วิธีแทนค่าว่างเปล่าด้วยค่าที่พบบ่อย
df['เพศ'].fillna(df['เพศ'].mode()[0], inplace=True)
df
```

```python
def standardize_gender(gender):
    gender = gender.lower()
    if gender in ['ญ','female','f']:
        return 'หญิง'
    elif gender in ['male','ช','m']:
        return 'ชาย'
    else:
        return gender

df['เพศ'] = df['เพศ'].apply(standardize_gender)
df
```

```python
#จัดการค่าผิดปกติ
Q1 = df['อายุ'].quantile(0.25)
Q3 = df['อายุ'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
df = df[(df['อายุ'] >= lower_bound) & (df['อายุ'] <= upper_bound)]
df
```

```python
df.loc[df['GPA'] < 0, 'GPA'] = np.nan
df.loc[df['GPA'] > 4, 'GPA'] = 4.0
mean_gpa = df['GPA'].mean()
df['GPA'].fillna(mean_gpa, inplace=True)
df['GPA']= df['GPA'].round(2)
df
```

```python
mean_weight = df['น้ำหนัก_kg'].mean()
df['น้ำหนัก_kg'] = df['น้ำหนัก_kg'].fillna(mean_weight).astype(int)
df
```

```python
mean_height = df['ส่วนสูง_cm'].mean()
df['ส่วนสูง_cm'] = df['ส่วนสูง_cm'].fillna(mean_height).astype(int)
df
```

```python
df_faculty = df[['คณะ', 'รหัสนักศึกษา']]
df_encoded = pd.get_dummies(df_faculty,columns=['คณะ'])
df_encoded
```
